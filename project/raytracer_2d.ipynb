# Raytracer algorithm

The raytracer simulates multiple lines originating from a point source limited to a certain opening angle. The rays of light only refract at interfaces, which are represented by straight or curved lines pieces which encapsulate different regions with different refractive indices. To prevent any issues with interfaces going towards infinity, we assume a square box around the simulation area that absorbs all rays at the walls of the box.
