## Ball motion

Iterative 

The motion will be updated while the ball is still moving fast enough or the gradient/slope is large enough. 
At each iteration both the position and the velocity of the ball are adapted:  

- the velocity of the ball is adapted by accelerating the ball towards the downhill gradient of the potential surface, i.e. the downhill slope. 
- the position of the ball is changed using the current velocity and acceleration:

$$
x_{j+1} = \frac{1}{2} a_j dt^2 + v_j dt + x_j
v_x = a_x t
$$

When the motion stops, the updates stop, and the fin 
There is also test whether the ball crossed a hole (circular area with zero slope), if it crossed, the velocity is set to zero, iterations are stopped and end-result is indicated hole with number something.

```python
# Creation of a surface

from numpy import meshgrid, sin, exp, cos, pi, sqrt, linspace, random
import matplotlib.pyplot as plt

Lx = 1; Ly = 1
x = linspace(0, Lx, 100)
y = linspace(0, Ly, 100)
xx, yy = meshgrid(x, y)

n_hill = 100; A_max = 5.0; s_max = 0.2
AA = random.uniform(low=-1, high=1.0, size=n_hill) * A_max
x0 = random.uniform(low=-s_max, high=1 + s_max, size=n_hill) * Lx
y0 = random.uniform(low=-s_max, high=1 + s_max, size=n_hill) * Ly
s0 = random.uniform(low=0.5, high=1.0, size=n_hill) * s_max
zz = 0 * xx; dxx = 0 * xx; dyy = 0 * xx
for j in range(n_hill):
    zz_hill = AA[j] * exp(- ((xx-x0[j]) ** 2 + (yy - y0[j]) ** 2) / (2 * s0[j] ** 2))
    zz = zz + zz_hill
    dxx = dxx - (xx-x0[j]) / s0[j] ** 2 * zz_hill 
    dyy = dyy - (yy-y0[j]) / s0[j] ** 2 * zz_hill
fig, axs = plt.subplots(1, 3)
ax = axs[0]
ax.pcolormesh(xx, yy, zz, cmap="jet")
ax.set_aspect("equal")
ax = axs[1]
ax.pcolormesh(xx, yy, dxx, cmap="jet")
ax.set_aspect("equal")
ax = axs[2]
ax.pcolormesh(xx, yy, dyy, cmap="jet")
ax.set_aspect("equal")
plt.show()
```
